# Notebook Example
Sphinx でノートブックをビルドする際の最小例です。上から順に実行し、出力を含めてドキュメント化できます。

```python
from sample_project.sample_add import sample_add

print("2 + 3 =", sample_add(2, 3))
```

## 複数の入力で実行する

ループや内包表記と組み合わせて、複数の入力に対する結果をまとめて表示できます。

```python
pairs = [(1, 2), (10, -3), (100, 200)]

for a, b in pairs:
    print(f"{a} + {b} = {sample_add(a, b)}")
```

## エラーハンドリングの例

`sample_add` は `int` 以外の引数に対して `TypeError` を送出します。ノートブックは Sphinx ビルド時に実行されるため、例外は `try/except` で捕捉して出力として残します。

```python
try:
    sample_add("2", 3)  # ty: ignore[invalid-argument-type]
except TypeError as e:
    print(f"caught: {e}")
```
